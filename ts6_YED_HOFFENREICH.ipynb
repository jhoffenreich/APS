# Desarrollo
Para el análisis de los distintos circuitos se siguió un procedimiento general basado en la obtención de la función de transferencia y el posterior estudio de su respuesta en frecuencia.

En primer lugar, los componentes reactivos fueron expresados mediante sus impedancias en el dominio de Laplace:

$$Z_R=R$$

$$Z_L=sL$$

$$Z_C=\frac{1}{sC}$$

A partir de las asociaciones serie y paralelo de cada circuito se determinaron las impedancias equivalentes necesarias para expresar la salida mediante un divisor de tensión:

$$H(s)=\frac{V_o(s)}{V_i(s)}=\frac{Z_2}{Z_1+Z_2}$$

donde $Z_2$ corresponde a la impedancia sobre la cual se mide la tensión de salida.

Una vez obtenida $H(s)$, se reordenó su expresión de forma que el denominador resultara mónico, es decir, con coeficiente unitario para el término de mayor grado. Al tratarse de sistemas de segundo orden, el denominador obtenido se comparó con la forma canónica:

$$s^2+\frac{\omega_0}{Q}s+\omega_0^2$$

Esta comparación permitió relacionar los parámetros $\omega_0$ y $Q$ establecidos en la consigna con los valores de los componentes $R$, $L$ y $C$ de cada circuito. Los valores obtenidos fueron posteriormente reemplazados en las funciones de transferencia correspondientes.

A continuación se determinaron los polos y ceros de cada sistema, calculando las raíces del denominador y del numerador de $H(s)$, respectivamente. Su ubicación en el plano complejo permitió complementar el análisis de estabilidad y anticipar características de la respuesta en frecuencia.

Para estudiar el comportamiento de cada filtro se evaluó la función de transferencia sobre el eje imaginario mediante:

$$s=j\omega$$

obteniendo así $H(j\omega)$. A partir de esta expresión se calcularon el módulo y la fase, y se estudió especialmente su comportamiento para:

$$\omega\rightarrow0$$

$$\omega=\omega_0$$

$$\omega\rightarrow\infty$$

Estos límites permitieron identificar si cada configuración correspondía a un filtro pasa bajos, pasa altos, pasa banda o elimina banda.

La respuesta asintótica también fue analizada a partir de los polos y ceros. Cada cero real de primer orden produce un incremento de pendiente de $+20\ dB/dec$, mientras que cada polo real de primer orden produce una disminución de $-20\ dB/dec$. En el caso de un par de polos o ceros complejos conjugados de segundo orden, la contribución total resulta de $\pm40\ dB/dec$. Los polos y ceros ubicados en el origen producen su contribución desde las bajas frecuencias, mientras que los restantes modifican la pendiente alrededor de su frecuencia característica.

Finalmente, se determinaron las frecuencias de corte a partir del criterio de $-3\ dB$ respecto de la ganancia de referencia de la banda pasante:

$$|H(j\omega_c)|=\frac{|H|_{ref}}{\sqrt2}$$

La elección de $|H|_{ref}$ depende del tipo de filtro. Para filtros pasa bajos se toma la ganancia a bajas frecuencias; para pasa altos, la ganancia a altas frecuencias; y para pasa banda o elimina banda se determinan dos frecuencias de corte, $\omega_1$ y $\omega_2$. A partir de estas frecuencias se obtuvo el ancho de banda correspondiente.

Las respuestas obtenidas analíticamente fueron verificadas numéricamente mediante Python, graficando los diagramas de módulo, fase y polos y ceros. ### Interpretación cualitativa de los componentes reactivos

Como criterio previo al desarrollo analítico, puede anticiparse el comportamiento de algunas configuraciones observando la ubicación de inductores y capacitores:

- **Inductor en serie:** tiende a producir un comportamiento pasa bajos, ya que su impedancia aumenta con la frecuencia.
- **Inductor en derivación:** tiende a producir un comportamiento pasa altos, ya que presenta baja impedancia a bajas frecuencias.
- **Capacitor en serie:** tiende a producir un comportamiento pasa altos, ya que su impedancia disminuye al aumentar la frecuencia.
- **Capacitor en derivación:** tiende a producir un comportamiento pasa bajos, desviando hacia masa las componentes de alta frecuencia.

Estas reglas permiten anticipar cualitativamente el comportamiento del circuito, aunque la clasificación definitiva se realizó a partir de la función de transferencia obtenida.

# Resultados y análisis

### a) Circuito RLC pasa bajos

Para obtener la función de transferencia del circuito se buscó inicialmente expresarlo como un divisor de tensión:

$$H(s)=\frac{Z_2}{Z_1+Z_2}$$

En este circuito, la resistencia y el inductor se encuentran en serie, por lo que:

$$Z_1=sL+R$$

mientras que la salida se mide sobre el capacitor:

$$Z_2=\frac{1}{sC}$$

Reemplazando ambas impedancias en el divisor de tensión:

$$H(s)=\frac{\frac{1}{sC}}{sL+R+\frac{1}{sC}}$$

Luego de simplificar para obtener un denominador mónico, se obtuvo:

$$H(s)=\frac{\frac{1}{LC}}{s^2+\frac{R}{L}s+\frac{1}{LC}}$$

El denominador se comparó con la forma canónica de un sistema de segundo orden:

$$s^2+\frac{\omega_0}{Q}s+\omega_0^2$$

De esta comparación se obtuvieron las relaciones:

$$\omega_0^2=\frac{1}{LC}$$

$$\frac{\omega_0}{Q}=\frac{R}{L}$$

Para el análisis se utilizaron los valores establecidos en la consigna:

$$\omega_0=1\ rad/s,\qquad R=1\ \Omega,\qquad Q=\frac{\sqrt2}{2}$$

Reemplazando dichos valores en las relaciones anteriores se obtuvo:

$$L=\frac{RQ}{\omega_0}=\frac{\sqrt2}{2}\ H$$

$$C=\frac{1}{L\omega_0^2}=\sqrt2\ F$$

Reemplazando estos valores en la función de transferencia:

$$\boxed{H(s)=\frac{1}{s^2+\sqrt2s+1}}$$

#### Polos y ceros

Como el numerador es constante, el sistema no presenta ceros finitos. Por otro lado, los polos se obtuvieron a partir de las raíces del denominador:

$$s^2+\sqrt2s+1=0$$

Aplicando la fórmula cuadrática:

$$s=\frac{-\sqrt2\pm\sqrt{2-4}}{2}$$

se obtuvieron:

$$p_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2}$$

$$p_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Ambos polos se encuentran en el semiplano izquierdo del plano complejo, por lo que el sistema resulta estable.

#### Respuesta en frecuencia

Para estudiar la respuesta en frecuencia se reemplazó $s=j\omega$, obteniéndose:

$$H(j\omega)=\frac{1}{1-\omega^2+j\sqrt2\omega}$$

El módulo de la función de transferencia resulta:

$$|H(j\omega)|=\frac{1}{\sqrt{(1-\omega^2)^2+2\omega^2}}=\frac{1}{\sqrt{1+\omega^4}}$$

y su valor en decibeles se calcula mediante:

$$|H(j\omega)|_{dB}=20\log_{10}|H(j\omega)|$$

Para bajas frecuencias, cuando $\omega\rightarrow0$:

$$|H(j\omega)|\rightarrow1$$

por lo tanto:

$$|H(j\omega)|_{dB}\rightarrow0\ dB$$

Esto indica que *las componentes de baja frecuencia atraviesan el sistema prácticamente sin atenuación*.

En la frecuencia natural, $\omega=\omega_0=1\ rad/s$:

$$|H(j1)|=\frac{1}{\sqrt2}$$

y en decibeles:

$$20\log_{10}\left(\frac{1}{\sqrt2}\right)=-3.01\ dB$$

Por lo tanto, para este caso particular, correspondiente a $Q=\frac{\sqrt2}{2}$, la frecuencia natural coincide con la *frecuencia de corte*:

$$\omega_c=\omega_0=1\ rad/s$$

Para frecuencias elevadas, cuando $\omega\rightarrow\infty$:

$$|H(j\omega)|\approx\frac{1}{\omega^2}\rightarrow0$$

por lo que:

$$|H(j\omega)|_{dB}\rightarrow-\infty$$

De esta manera, el circuito permite el paso de bajas frecuencias y atenúa las altas frecuencias, por lo que se identifica como un **filtro pasa bajos de segundo orden**.

Como se vio en clase, este comportamiento podía anticiparse cualitativamente por la presencia de un inductor en serie y un capacitor en derivación, ambos favoreciendo el paso de bajas frecuencias.

#### Fase

La fase de la función de transferencia se obtiene como la diferencia entre la fase del numerador y la del denominador. Como el numerador es real y positivo:

$$\phi(\omega)=-\arg(1-\omega^2+j\sqrt2\omega)$$

En forma general puede expresarse como:

$$\phi(\omega)=-\tan^{-1}\left(\frac{\sqrt2\omega}{1-\omega^2}\right)$$

teniendo en cuenta el cuadrante correspondiente.

Para $\omega\rightarrow0$, el denominador tiende a un número real positivo, por lo que:

$$\phi(\omega)\rightarrow0^\circ$$

Para $\omega=\omega_0=1\ rad/s$:

$$H(j1)=\frac{1}{j\sqrt2}$$

El denominador se encuentra sobre el eje imaginario positivo y presenta una fase de $90^\circ$. Por lo tanto:

$$\phi(1)=-90^\circ$$

Finalmente, cuando $\omega\rightarrow\infty$, domina el término $-\omega^2$ del denominador, correspondiente a un número real negativo con fase de $180^\circ$. Por lo tanto:

$$\phi(\omega)\rightarrow-180^\circ$$

#### Respuesta asintótica

El sistema posee dos polos y no presenta ceros finitos. Cada polo aporta una pendiente de $-20\ dB/dec$ una vez alcanzada su región de influencia. Por lo tanto, la pendiente asintótica para altas frecuencias resulta:

$$-40\ dB/dec$$

La respuesta de módulo puede resumirse entonces como:

- Para $\omega\ll1\ rad/s$: $|H|_{dB}\approx0\ dB$.
- Para $\omega=1\ rad/s$: $|H|_{dB}=-3.01\ dB$.
- Para $\omega\gg1\ rad/s$: la respuesta cae con una pendiente aproximada de $-40\ dB/dec$.

#### Frecuencia de corte y ancho de banda

Para un filtro pasa bajos, la frecuencia de corte se determina a partir de:

$$|H(j\omega_c)|=\frac{|H(0)|}{\sqrt2}$$

Como $|H(0)|=1$:

$$\frac{1}{\sqrt{1+\omega_c^4}}=\frac{1}{\sqrt2}$$

De esta expresión se obtiene:

$$\omega_c^4=1$$

y considerando frecuencias positivas:

$$\omega_c=1\ rad/s$$

La banda pasante de un filtro pasa bajos se extiende desde $\omega=0$ hasta la frecuencia de corte. Por lo tanto:

$$BW=\omega_c-0=1\ rad/s$$

Expresando la frecuencia de corte en hertz:

$$f_c=\frac{\omega_c}{2\pi}=\frac{1}{2\pi}\approx0.159\ Hz$$

Posteriormente, estas respuestas fueron simuladas numéricamente mediante Python con el objetivo de comparar los resultados obtenidos analíticamente con los gráficos generados computacionalmente. Se observó una buena concordancia entre ambos métodos. El código utilizado para realizar la simulación y generar los gráficos se presenta a continuación:

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.signal as signal

# =============================================================================
# Parámetros del circuito a)
# =============================================================================
w0 = 1.0
Q = np.sqrt(2)/2

# H(s) = 1 / (s^2 + sqrt(2)s + 1)
num = [1]
den = [1, np.sqrt(2), 1]

sys_a = signal.TransferFunction(num, den)

# =============================================================================
# Respuesta en frecuencia
# =============================================================================

# Solo frecuencias positivas
w = np.logspace(-2, 2, 1000)

w_a, mag_a, phase_a = signal.bode(sys_a, w)

# Respuesta asintótica del módulo:
# 0 dB hasta w0 y -40 dB/dec para w > w0
asintota_mag = np.where(
    w <= w0,
    0,
    -40*np.log10(w/w0)
)

# Nivel de corte exacto
nivel_corte = 20*np.log10(1/np.sqrt(2))

# Polos y ceros
polos = sys_a.poles
ceros = sys_a.zeros

# =============================================================================
# Gráficos
# =============================================================================

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# -----------------------------------------------------------------------------
# 1. Módulo
# -----------------------------------------------------------------------------

ax[0].semilogx(
    w_a,
    mag_a,
    linewidth=2,
    label='Respuesta de módulo'
)

ax[0].semilogx(
    w,
    asintota_mag,
    '--',
    linewidth=1.5,
    label='Respuesta asintótica del módulo'
)

# Frecuencia de corte
ax[0].axvline(
    w0,
    color='tab:red',
    linestyle=':',
    label=r'$\omega_c=\omega_0=1$ rad/s'
)

# Nivel de -3.01 dB
ax[0].axhline(
    nivel_corte,
    color='tab:green',
    linestyle=':',
    label=r'Nivel de corte: $-3.01$ dB'
)

# Punto de corte
ax[0].plot(
    w0,
    nivel_corte,
    'o',
    color='tab:green'
)

# Pendiente asintótica
ax[0].annotate(
    r'$-40$ dB/dec',
    xy=(10, -40),
    xytext=(10, 8),
    textcoords='offset points',
    fontsize=10
)

ax[0].set_title('Módulo - Circuito a) Pasa bajos')
ax[0].set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax[0].set_ylabel(r'$|H(j\omega)|$ [dB]')
ax[0].grid(True, which='both', linestyle='--', alpha=0.7)
ax[0].legend(loc='lower left')


# -----------------------------------------------------------------------------
# 2. Fase
# -----------------------------------------------------------------------------

ax[1].semilogx(
    w_a,
    phase_a,
    linewidth=2,
    label='Respuesta de fase'
)

ax[1].axvline(
    w0,
    color='tab:red',
    linestyle=':',
    label=r'$\omega_0=1$ rad/s'
)

ax[1].axhline(
    -90,
    color='tab:green',
    linestyle=':',
    label=r'$\phi(\omega_0)=-90^\circ$'
)

ax[1].plot(
    w0,
    -90,
    'o',
    color='tab:green'
)

ax[1].set_title('Fase - Circuito a) Pasa bajos')
ax[1].set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax[1].set_ylabel(r'Fase $\phi(\omega)$ [$^\circ$]')
ax[1].grid(True, which='both', linestyle='--', alpha=0.7)
ax[1].legend(loc='lower left')


# -----------------------------------------------------------------------------
# 3. Polos y ceros
# -----------------------------------------------------------------------------

theta = np.linspace(0, 2*np.pi, 200)

# Circunferencia de radio w0
ax[2].plot(
    w0*np.cos(theta),
    w0*np.sin(theta),
    ':',
    label=r'$|s|=\omega_0=1$'
)

# Ejes del plano complejo
ax[2].axhline(0, color='black', linewidth=1)
ax[2].axvline(0, color='black', linewidth=1)

# Polos
ax[2].plot(
    np.real(polos),
    np.imag(polos),
    'x',
    markersize=10,
    markeredgewidth=2,
    label='Polos'
)

# Ceros
if len(ceros) > 0:
    ax[2].plot(
        np.real(ceros),
        np.imag(ceros),
        'o',
        markersize=8,
        markerfacecolor='none',
        label='Ceros'
    )

ax[2].set_title('Diagrama de polos y ceros')
ax[2].set_xlabel('Parte real')
ax[2].set_ylabel('Parte imaginaria')
ax[2].grid(True, linestyle='--', alpha=0.7)
ax[2].axis('equal')
ax[2].legend(loc='upper right')

plt.tight_layout()
plt.show()
```

### b) Circuito RLC pasa altos

Para obtener la función de transferencia del circuito se buscó nuevamente expresarlo como un divisor de tensión:

$$H(s)=\frac{Z_2}{Z_1+Z_2}$$

En este caso, la resistencia y el capacitor se encuentran en paralelo, por lo que la impedancia equivalente resulta:

$$Z_1=Z_R\parallel Z_C=\frac{R}{1+sCR}$$

mientras que la salida se mide sobre el inductor:

$$Z_2=sL$$

Reemplazando ambas impedancias en el divisor de tensión:

$$H(s)=\frac{sL}{\frac{R}{1+sCR}+sL}$$

Luego de simplificar para obtener un denominador mónico, se obtuvo:

$$H(s)=\frac{s^2+\frac{1}{CR}s}{s^2+\frac{1}{CR}s+\frac{1}{CL}}$$

El denominador se comparó con la forma canónica de un sistema de segundo orden:

$$s^2+\frac{\omega_0}{Q}s+\omega_0^2$$

De esta comparación se obtuvieron las relaciones:

$$\frac{\omega_0}{Q}=\frac{1}{CR}$$

$$\omega_0^2=\frac{1}{CL}$$

Para el análisis se utilizaron nuevamente los valores establecidos en la consigna:

$$\omega_0=1\ rad/s,\qquad R=1\ \Omega,\qquad Q=\frac{\sqrt2}{2}$$

Reemplazando dichos valores en las relaciones anteriores se obtuvo:

$$C=\frac{Q}{\omega_0R}=\frac{\sqrt2}{2}\ F$$

$$L=\frac{1}{C\omega_0^2}=\sqrt2\ H$$

Reemplazando estos valores en la función de transferencia:

$$\boxed{H(s)=\frac{s^2+\sqrt2s}{s^2+\sqrt2s+1}}$$

#### Polos y ceros

Los ceros se obtuvieron a partir de las raíces del numerador:

$$s^2+\sqrt2s=0$$

Factorizando:

$$s(s+\sqrt2)=0$$

por lo que:

$$z_1=0$$

$$z_2=-\sqrt2$$

Los polos se obtuvieron a partir de las raíces del denominador. Como este coincide con el obtenido para el circuito a), se obtuvieron los mismos polos:

$$p_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2}$$

$$p_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Ambos polos se encuentran en el semiplano izquierdo del plano complejo, por lo que el sistema resulta estable.

#### Respuesta en frecuencia

Para estudiar la respuesta en frecuencia se reemplazó $s=j\omega$, obteniéndose:

$$H(j\omega)=\frac{-\omega^2+j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega}$$

El módulo de la función de transferencia resulta:

$$|H(j\omega)|=\frac{\sqrt{\omega^4+2\omega^2}}{\sqrt{(1-\omega^2)^2+2\omega^2}}$$

Como se consideran frecuencias positivas:

$$|H(j\omega)|=\frac{\omega\sqrt{\omega^2+2}}{\sqrt{1+\omega^4}}$$

y su valor en decibeles se calcula mediante:

$$|H(j\omega)|_{dB}=20\log_{10}|H(j\omega)|$$

Para bajas frecuencias, cuando $\omega\rightarrow0^+$:

$$|H(j\omega)|\rightarrow0$$

por lo que:

$$|H(j\omega)|_{dB}\rightarrow-\infty$$

Esto indica que las componentes de baja frecuencia son fuertemente atenuadas.

Para frecuencias elevadas, cuando $\omega\rightarrow\infty$, los términos de mayor orden dominan tanto en el numerador como en el denominador:

$$H(j\omega)\approx\frac{-\omega^2}{-\omega^2}=1$$

por lo tanto:

$$|H(j\omega)|\rightarrow1$$

y:

$$|H(j\omega)|_{dB}\rightarrow0\ dB$$

De esta manera, el circuito atenúa las bajas frecuencias y permite el paso de las altas frecuencias, por lo que se identifica como un **filtro pasa altos de segundo orden**.
Como se vio en clase, este comportamiento podía anticiparse cualitativamente por la presencia del capacitor en la rama de entrada y del inductor en derivación, configuración que favorece el paso de altas frecuencias.

En la frecuencia natural, $\omega=\omega_0=1\ rad/s$:

$$|H(j1)|=\frac{\sqrt3}{\sqrt2}=\sqrt{\frac32}$$

y en decibeles:

$$20\log_{10}\left(\sqrt{\frac32}\right)\approx1.76\ dB$$

Por lo tanto, en $\omega=\omega_0$ la respuesta presenta una ganancia superior a $0\ dB$.

#### Fase

La fase de la función de transferencia se obtiene como la diferencia entre la fase del numerador y la del denominador:

$$\phi(\omega)=\arg(-\omega^2+j\sqrt2\omega)-\arg(1-\omega^2+j\sqrt2\omega)$$

Para $\omega\rightarrow0^+$:

$$H(j\omega)\approx j\sqrt2\omega$$

por lo que la respuesta se aproxima al origen a través del eje imaginario positivo y:

$$\phi(\omega)\rightarrow90^\circ$$

En $\omega=\omega_0=1\ rad/s$:

$$H(j1)=\frac{-1+j\sqrt2}{j\sqrt2}$$

La fase del numerador es aproximadamente $125^\circ$, mientras que la fase del denominador es $90^\circ$. Por lo tanto:

$$\phi(1)\approx35^\circ$$

Finalmente, cuando $\omega\rightarrow\infty$:

$$H(j\omega)\approx1$$

por lo que:

$$\phi(\omega)\rightarrow0^\circ$$

#### Respuesta asintótica

El sistema posee dos ceros y dos polos. El cero ubicado en el origen aporta una pendiente de:

$$+20\ dB/dec$$

desde las bajas frecuencias.

El segundo cero se encuentra en:

$$z_2=-\sqrt2$$

por lo que alrededor de:

$$\omega_z=\sqrt2\ rad/s$$

agrega otros:

$$+20\ dB/dec$$

Por otro lado, el par de polos complejos conjugados presenta una frecuencia natural $\omega_0=1\ rad/s$ y aporta en conjunto:

$$-40\ dB/dec$$

De esta manera, la pendiente asintótica resulta aproximadamente:

- Para $\omega<1\ rad/s$: $+20\ dB/dec$.
- Para $1<\omega<\sqrt2\ rad/s$: $-20\ dB/dec$.
- Para $\omega>\sqrt2\ rad/s$: $0\ dB/dec$.

A altas frecuencias, las contribuciones de polos y ceros se compensan:

$$+20+20-40=0\ dB/dec$$

Esto coincide con el comportamiento obtenido analíticamente, ya que para frecuencias elevadas la respuesta tiende a una ganancia constante de $0\ dB$.

La respuesta de módulo puede resumirse entonces como:

- Para $\omega\rightarrow0^+$: $|H|_{dB}\rightarrow-\infty$.
- Para $\omega=1\ rad/s$: $|H|_{dB}\approx1.76\ dB$.
- Para $\omega\rightarrow\infty$: $|H|_{dB}\rightarrow0\ dB$.

#### Frecuencia de corte y ancho de banda

Para un filtro pasa altos, la frecuencia de corte se determina tomando como referencia la ganancia de la banda pasante a altas frecuencias:

$$|H(j\omega_c)|=\frac{|H(\infty)|}{\sqrt2}$$

Como:

$$|H(\infty)|=1$$

se obtiene:

$$|H(j\omega_c)|=\frac{1}{\sqrt2}$$

Reemplazando la expresión del módulo:

$$\frac{\omega_c\sqrt{\omega_c^2+2}}{\sqrt{1+\omega_c^4}}=\frac{1}{\sqrt2}$$

Elevando ambos miembros al cuadrado y simplificando:

$$\omega_c^4+4\omega_c^2-1=0$$

Para resolver esta ecuación se realizó el cambio de variable:

$$x=\omega_c^2$$

obteniéndose:

$$x^2+4x-1=0$$

Las soluciones son:

$$x_1=-2+\sqrt5$$

$$x_2=-2-\sqrt5$$

La segunda solución se descarta porque $x=\omega_c^2$ debe ser positivo. Por lo tanto:

$$\omega_c^2=-2+\sqrt5$$

y finalmente:

$$\boxed{\omega_c=\sqrt{-2+\sqrt5}\approx0.486\ rad/s}$$

Expresando la frecuencia de corte en hertz:

$$f_c=\frac{\omega_c}{2\pi}\approx0.0774\ Hz$$

La banda pasante de un filtro pasa altos se extiende desde la frecuencia de corte hasta frecuencias arbitrariamente elevadas:

$$\omega_c<\omega<\infty$$

Por lo tanto:

$$BW=\omega_{max}-\omega_c$$

y como:

$$\omega_{max}\rightarrow\infty$$

se concluye que:

$$\boxed{BW\rightarrow\infty}$$

Por lo tanto, el filtro pasa altos ideal no presenta un ancho de banda finito.

### c) Circuito RLC pasa banda

En este circuito, la resistencia se encuentra en serie con la asociación en paralelo del inductor y el capacitor. Por lo tanto:

$$Z_1=R$$

$$Z_2=Z_L\parallel Z_C=\frac{sL}{1+s^2LC}$$

Aplicando el divisor de tensión:

$$H(s)=\frac{Z_2}{Z_1+Z_2}
=\frac{\frac{sL}{1+s^2LC}}{R+\frac{sL}{1+s^2LC}}$$

Reordenando la expresión y buscando un denominador mónico:

$$H(s)=\frac{\frac{s}{RC}}{s^2+\frac{1}{RC}s+\frac{1}{LC}}$$

Comparando el denominador con la forma canónica:

$$s^2+\frac{\omega_0}{Q}s+\omega_0^2$$

se obtienen nuevamente las relaciones:

$$\frac{\omega_0}{Q}=\frac{1}{RC}$$

$$\omega_0^2=\frac{1}{LC}$$

Por lo tanto, para:

$$\omega_0=1\ rad/s,\qquad R=1\ \Omega,\qquad Q=\frac{\sqrt2}{2}$$

se obtienen los mismos valores de componentes que en el circuito b):

$$C=\frac{\sqrt2}{2}\ F$$

$$L=\sqrt2\ H$$

Reemplazando estos valores:

$$\boxed{H(s)=\frac{\sqrt2s}{s^2+\sqrt2s+1}}$$

#### Polos y ceros

El único cero finito se obtiene a partir del numerador:

$$\sqrt2s=0$$

por lo que:

$$z_1=0$$

El denominador coincide nuevamente con el de los circuitos anteriores, por lo que los polos son:

$$p_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2}$$

$$p_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Ambos polos se encuentran en el semiplano izquierdo, por lo que el sistema resulta estable.

#### Respuesta en frecuencia

Reemplazando $s=j\omega$:

$$H(j\omega)=\frac{j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega}$$

El módulo resulta:

$$|H(j\omega)|=
\frac{\sqrt2\omega}{\sqrt{(1-\omega^2)^2+2\omega^2}}
=
\frac{\sqrt2\omega}{\sqrt{1+\omega^4}}$$

El comportamiento del módulo y la fase puede analizarse directamente para bajas frecuencias, en la frecuencia natural y para altas frecuencias:

| Frecuencia | Módulo | Módulo [dB] | Fase |
|---|---|---|---|
| $\omega\rightarrow0^+$ | $|H|\rightarrow0$ | $-\infty\ dB$ | $90^\circ$ |
| $\omega=\omega_0=1\ rad/s$ | $|H|=1$ | $0\ dB$ | $0^\circ$ |
| $\omega\rightarrow\infty$ | $|H|\rightarrow0$ | $-\infty\ dB$ | $-90^\circ$ |

Para bajas frecuencias, cuando $\omega\rightarrow0^+$:

$$H(j\omega)\approx j\sqrt2\omega$$

Como $\omega$ es positiva, esta expresión corresponde a un número imaginario positivo. Por lo tanto, su argumento tiende a:

$$\phi(\omega)\rightarrow90^\circ$$

Además, el módulo tiende a cero:

$$|H(j\omega)|\rightarrow0$$

por lo que, expresado en decibeles:

$$|H(j\omega)|_{dB}\rightarrow-\infty$$

En la frecuencia natural, $\omega=\omega_0=1\ rad/s$:

$$H(j1)=\frac{j\sqrt2}{-1+j\sqrt2+1}
=\frac{j\sqrt2}{j\sqrt2}=1$$

Por lo tanto:

$$|H(j1)|=1\qquad\Rightarrow\qquad |H(j1)|_{dB}=0\ dB$$

Como $H(j1)=1$ es un número real positivo, su argumento es nulo y, en consecuencia:

$$\phi(1)=0^\circ$$

Finalmente, para frecuencias elevadas, cuando $\omega\rightarrow\infty$, dominan los términos de mayor orden:

$$H(j\omega)\approx\frac{j\sqrt2\omega}{-\omega^2}
=-j\frac{\sqrt2}{\omega}$$

Como $\omega$ tiende a infinito:

$$|H(j\omega)|\rightarrow0$$

y por lo tanto:

$$|H(j\omega)|_{dB}\rightarrow-\infty$$

Además, la expresión aproximada corresponde a un número imaginario negativo, por lo que su argumento tiende a:

$$\phi(\omega)\rightarrow-90^\circ$$

El circuito atenúa tanto las bajas como las altas frecuencias, mientras que presenta su máxima ganancia en torno a $\omega_0$. Por lo tanto, se identifica como un **filtro pasa banda de segundo orden**.
Como se vio en clase, este comportamiento podía anticiparse cualitativamente por la combinación en derivación del capacitor, asociado a un comportamiento pasa bajos, y del inductor, asociado a un comportamiento pasa altos.
#### Respuesta asintótica

A partir de los polos y ceros del sistema puede justificarse la pendiente del diagrama de Bode de módulo.

La función de transferencia es:

$$H(s)=\frac{\sqrt2s}{s^2+\sqrt2s+1}$$

El sistema presenta un cero en el origen, el cual aporta una pendiente de: $+20\ dB/dec$ desde las bajas frecuencias. Por este motivo, para $\omega\ll\omega_0$ el módulo aumenta aproximadamente con una pendiente de $+20\ dB/dec$.

Por otro lado, el sistema posee dos polos complejos conjugados asociados a $\omega_0=1\ rad/s$. En conjunto, estos polos aportan: $-40\ dB/dec$  a partir de la región de la frecuencia natural.

Como ya existía una contribución de $+20\ dB/dec$ debida al cero en el origen, para frecuencias superiores a $\omega_0$ la pendiente neta resulta:

$$+20-40=-20\ dB/dec$$

Por lo tanto, la respuesta asintótica del módulo presenta una pendiente de:

$$\boxed{+20\ dB/dec\quad \text{para bajas frecuencias}}$$

y:

$$\boxed{-20\ dB/dec\quad \text{para altas frecuencias}}$$

Este comportamiento concuerda con el carácter pasa banda del circuito: el módulo aumenta desde bajas frecuencias hasta alcanzar su máximo alrededor de $\omega_0$ y luego disminuye para frecuencias superiores.

#### Frecuencias de corte y ancho de banda

Al tratarse de un filtro pasa banda existen dos frecuencias de corte, $\omega_1$ y $\omega_2$. Como el valor máximo del módulo es:

$$|H|_{max}=1$$

las frecuencias de corte corresponden a los puntos de $-3.01\ dB$:

$$|H(j\omega_c)|=\frac{1}{\sqrt2}$$

Reemplazando la expresión del módulo:

$$\frac{\sqrt2\omega_c}{\sqrt{1+\omega_c^4}}=\frac{1}{\sqrt2}$$

Elevando al cuadrado y simplificando:

$$\omega_c^4-4\omega_c^2+1=0$$

Realizando el cambio de variable:

$$x=\omega_c^2$$

se obtiene:

$$x^2-4x+1=0$$

cuyas soluciones son:

$$x_1=2-\sqrt3$$

$$x_2=2+\sqrt3$$

Por lo tanto, las dos frecuencias de corte son:

$$\boxed{\omega_1=\sqrt{2-\sqrt3}\approx0.518\ rad/s}$$

$$\boxed{\omega_2=\sqrt{2+\sqrt3}\approx1.932\ rad/s}$$

El ancho de banda queda determinado por:

$$BW=\omega_2-\omega_1$$

por lo que:

$$BW=\sqrt{2+\sqrt3}-\sqrt{2-\sqrt3}$$

y simplificando:

$$\boxed{BW=\sqrt2\approx1.414\ rad/s}$$

Este resultado coincide con la relación característica de un filtro pasa banda de segundo orden:

$$BW=\frac{\omega_0}{Q}=\frac{1}{\frac{\sqrt2}{2}}=\sqrt2\ rad/s$$

## Bonus
Para analizar el circuito se lo expresó nuevamente como un divisor de tensión. Para ello, se determinaron las siguientes impedancias equivalentes. La impedancia ubicada entre la entrada y la salida corresponde al paralelo del inductor y el capacitor:

$$Z_1=Z_L\parallel Z_C=\frac{sL}{aLCs^2+b}$$

Por otro lado, la impedancia conectada entre la salida y tierra corresponde al paralelo del capacitor, el inductor y la resistencia:

$$Z_2=Z_C\parallel Z_L\parallel R =\frac{sL}{LC(1-a)s^2+\frac{L}{R}s+(1-b)}$$

La función de transferencia puede escribirse mediante el divisor de tensión:

$$H(s)=\frac{Z_2}{Z_1+Z_2}$$

Para simplificar el desarrollo se definieron:

$$\mathcal{A}=aLCs^2+b$$

$$\mathcal{B}=LC(1-a)s^2+\frac{L}{R}s+(1-b)$$

De esta manera:

$$Z_1=\frac{sL}{\mathcal{A}}$$

$$Z_2=\frac{sL}{\mathcal{B}}$$

y la función de transferencia queda:

$$H(s)=\frac{\mathcal{A}}{\mathcal{A}+\mathcal{B}}$$

Reemplazando ambas expresiones:

$$H(s)=
\frac{aLCs^2+b}
{LCs^2+\frac{L}{R}s+1}$$

Dividiendo numerador y denominador por $LC$:

$$H(s)=
\frac{as^2+\frac{b}{LC}}
{s^2+\frac{1}{RC}s+\frac{1}{LC}}$$

El denominador se comparó nuevamente con la forma canónica:

$$s^2+\frac{\omega_0}{Q}s+\omega_0^2$$

por lo que:

$$\frac{\omega_0}{Q}=\frac{1}{RC}$$

$$\omega_0^2=\frac{1}{LC}$$

Para:

$$\omega_0=1\ rad/s,\qquad R=1\ \Omega,\qquad Q=\frac{\sqrt2}{2}$$

se obtienen:

$$C=\frac{Q}{\omega_0R}=\frac{\sqrt2}{2}\ F$$

$$L=\frac{1}{C\omega_0^2}=\sqrt2\ H$$

Por lo tanto, la función de transferencia general normalizada queda:

$$\boxed{H(s)=\frac{as^2+b}{s^2+\sqrt2s+1}}$$

#### Polos y ceros

El denominador coincide con el obtenido en los circuitos anteriores, por lo que los polos se mantienen en:

$$p_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2}$$

$$p_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Los ceros, en cambio, dependen de los parámetros $a$ y $b$. Igualando el numerador a cero:

$$as^2+b=0$$

se obtiene:

$$s^2=-\frac{b}{a}$$

y por lo tanto:

$$z_{1,2}=\pm j\sqrt{\frac{b}{a}}$$

En forma general, la frecuencia angular asociada a los ceros es:

$$\omega_z=\omega_0\sqrt{\frac{b}{a}}$$

por lo que:

$$z_{1,2}=\pm j\omega_z$$

Como ambos ceros se encuentran sobre el eje imaginario, para $\omega=\omega_z$ el numerador de la función de transferencia se anula:

$$H(j\omega_z)=0$$

Por lo tanto:

$$|H(j\omega_z)|=0$$

y en decibeles:

$$|H(j\omega_z)|_{dB}\rightarrow-\infty$$

Esta frecuencia es completamente rechazada por el filtro. Además, los parámetros $a$ y $b$ permiten modificar la ubicación de dicha frecuencia.

Si se elige:

$$a=b$$

entonces:

$$\omega_z=\omega_0\sqrt{\frac{a}{a}}=\omega_0$$

Por lo tanto, la frecuencia rechazada coincide con la frecuencia natural del sistema.

Para este análisis se seleccionó:

$$a=b=1$$

de modo que:

$$\omega_z=\omega_0=1\ rad/s$$

y la función de transferencia queda:

$$\boxed{H(s)=\frac{s^2+1}{s^2+\sqrt2s+1}}$$

En este caso los ceros son:

$$z_1=+j$$

$$z_2=-j$$

mientras que los polos continúan siendo:

$$p_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2}$$

$$p_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

#### Respuesta en frecuencia

Reemplazando:

$$s=j\omega$$

se obtiene:

$$H(j\omega)=
\frac{1-\omega^2}
{1-\omega^2+j\sqrt2\omega}$$

El módulo resulta:

$$|H(j\omega)|=
\frac{|1-\omega^2|}
{\sqrt{(1-\omega^2)^2+2\omega^2}}
=
\frac{|1-\omega^2|}
{\sqrt{1+\omega^4}}$$

El comportamiento del sistema puede resumirse de la siguiente manera:

| Frecuencia | Módulo [dB] | Fase |
|---|---:|---:|
| $\omega\rightarrow0$ | $0\ dB$ | $0^\circ$ |
| $\omega=\omega_0=1\ rad/s$ | $-\infty\ dB$ | No definida |
| $\omega\rightarrow\infty$ | $0\ dB$ | $0^\circ$ |

Para bajas frecuencias:

$$H(0)=1$$

por lo que:

$$|H(0)|=1\qquad\Rightarrow\qquad |H(0)|_{dB}=0\ dB$$

Además, al tratarse de un número real positivo:

$$\phi(0)=0^\circ$$

Por lo tanto, las bajas frecuencias atraviesan el sistema prácticamente sin atenuación.

Para $\omega=\omega_0=1\ rad/s$:

$$H(j1)=0$$

por lo que:

$$|H(j1)|=0$$

y:

$$|H(j1)|_{dB}\rightarrow-\infty$$

En este punto la fase no se encuentra definida, debido a que la función de transferencia es exactamente nula.

Finalmente, para frecuencias elevadas:

$$H(j\omega)\approx
\frac{-\omega^2}{-\omega^2}=1$$

por lo que:

$$|H(j\omega)|\rightarrow1$$

$$|H(j\omega)|_{dB}\rightarrow0\ dB$$

y:

$$\phi(\omega)\rightarrow0^\circ$$

Por lo tanto, el circuito permite el paso de las frecuencias bajas y altas, mientras que rechaza completamente una frecuencia intermedia. Se identifica entonces como un **filtro notch o elimina banda**.

#### Comportamiento de la fase alrededor de la frecuencia de rechazo

Aunque la fase no está definida exactamente en $\omega=\omega_0$, puede analizarse su comportamiento a ambos lados de dicha frecuencia.

Para frecuencias apenas menores que $\omega_0$:

$$\omega<1$$

el numerador:

$$1-\omega^2>0$$

es un número real positivo, cuya fase es aproximadamente $0^\circ$ .

Al mismo tiempo, cuando $\omega\rightarrow1^-$, la parte real del denominador tiende a cero por valores positivos, mientras que su parte imaginaria permanece positiva. Por lo tanto, el denominador se aproxima al eje imaginario positivo y su fase tiende a $90^\circ$.

De esta manera:

$$\phi(\omega)\rightarrow0^\circ-90^\circ=-90^\circ$$

por lo que:

$$\boxed{\phi(\omega)\rightarrow-90^\circ\qquad\text{cuando}\qquad\omega\rightarrow\omega_0^-}$$

Para frecuencias apenas superiores a $\omega_0$:

$$\omega>1$$

el numerador:

$$1-\omega^2<0$$

es un número real negativo, cuya fase es $180^\circ$.

Por otro lado, la parte real del denominador es negativa y cercana a cero, mientras que su parte imaginaria continúa siendo positiva. El denominador se encuentra entonces en el segundo cuadrante y su fase tiende nuevamente a $90^\circ$.

Por lo tanto:

$$\phi(\omega)\rightarrow180^\circ-90^\circ=90^\circ$$

y:

$$\boxed{\phi(\omega)\rightarrow+90^\circ\qquad\text{cuando}\qquad\omega\rightarrow\omega_0^+}$$

Así, la fase presenta una discontinuidad al atravesar la frecuencia de rechazo: se aproxima a $-90^\circ$ antes del notch y a $+90^\circ$ inmediatamente después.

#### Frecuencias de corte y ancho de banda rechazado

Al tratarse de un filtro notch existen dos frecuencias de corte que delimitan la banda rechazada.

La ganancia de referencia en las bandas pasantes es:

$$|H|_{ref}=1$$

Por lo tanto, los puntos de corte de $-3.01\ dB$ cumplen:

$$|H(j\omega_c)|=\frac{1}{\sqrt2}$$

Reemplazando el módulo:

$$\frac{|1-\omega_c^2|}
{\sqrt{1+\omega_c^4}}
=\frac{1}{\sqrt2}$$

Elevando al cuadrado:

$$\frac{(1-\omega_c^2)^2}
{1+\omega_c^4}
=\frac12$$

y simplificando:

$$\omega_c^4-4\omega_c^2+1=0$$

Realizando el cambio de variable:

$$x=\omega_c^2$$

se obtiene:

$$x^2-4x+1=0$$

cuyas soluciones son:

$$x_1=2+\sqrt3$$

$$x_2=2-\sqrt3$$

Por lo tanto:

$$\boxed{\omega_1=\sqrt{2+\sqrt3}\approx1.932\ rad/s}$$

$$\boxed{\omega_2=\sqrt{2-\sqrt3}\approx0.518\ rad/s}$$

Estas frecuencias coinciden con las obtenidas para el filtro pasa banda del inciso anterior.

En ambos puntos:

$$|H(j\omega_1)|_{dB}=|H(j\omega_2)|_{dB}=-3.01\ dB$$

El ancho de la banda rechazada se obtiene como:

$$BW=\omega_1-\omega_2$$

por lo tanto:

$$BW=
\sqrt{2+\sqrt3}
-\sqrt{2-\sqrt3}$$

y simplificando:

$$\boxed{BW=\sqrt2\approx1.414\ rad/s}$$

Este resultado también coincide con:

$$BW=\frac{\omega_0}{Q}
=\frac{1}{\frac{\sqrt2}{2}}
=\sqrt2\ rad/s$$

Posteriormente, estas respuestas fueron simuladas numéricamente mediante Python con el objetivo de verificar los resultados obtenidos analíticamente. El código utilizado para realizar la simulación y generar los gráficos se presenta a continuación:

```python
#%% CIRCUITO BONUS A=B=1

import matplotlib.pyplot as plt
import numpy as np
import scipy.signal as signal

# =============================================================================
# 1. Parámetros - Filtro elimina banda / Notch
# =============================================================================
w0 = 1.0
Q = np.sqrt(2) / 2

# Para a = b = 1:
# H(s) = (s^2 + 1) / (s^2 + sqrt(2)s + 1)

num = [1, 0, 1]
den = [1, np.sqrt(2), 1]

sys_notch = signal.TransferFunction(num, den)

# Frecuencias de corte calculadas analíticamente
w1 = np.sqrt(2 - np.sqrt(3))
w2 = np.sqrt(2 + np.sqrt(3))

# =============================================================================
# 2. Respuesta en frecuencia
# =============================================================================

# Solo frecuencias positivas
w = np.logspace(-2, 2, 2000)

# Evaluación directa de H(jw)
s = 1j*w
H = (s**2 + 1)/(s**2 + np.sqrt(2)*s + 1)

# Módulo en dB
mag = 20*np.log10(np.abs(H))

# Fase
phase = np.angle(H, deg=True)

# Asíntota de módulo:
# el filtro tiende a 0 dB tanto para bajas como para altas frecuencias
asintota_mag = np.zeros_like(w)

# Nivel de corte
nivel_corte = 20*np.log10(1/np.sqrt(2))

# =============================================================================
# 3. Gráficos
# =============================================================================
fig = plt.figure(figsize=(15, 5))

# -----------------------------------------------------------------------------
# PANEL 1: Módulo
# -----------------------------------------------------------------------------
ax1 = plt.subplot(1, 3, 1)

ax1.semilogx(w, mag, 'b-', linewidth=2,
             label='Respuesta real')

ax1.semilogx(w, asintota_mag, 'k--', linewidth=1.5,
             alpha=0.8, label='Asíntota: 0 dB')

# Nivel de -3 dB
ax1.axhline(nivel_corte, color='r', linestyle=':',
            label=r'Nivel de corte: $-3.01$ dB')

# Frecuencias de corte
ax1.axvline(w1, color='g', linestyle=':',
            label=rf'$\omega_1={w1:.3f}$ rad/s')

ax1.axvline(w2, color='orange', linestyle=':',
            label=rf'$\omega_2={w2:.3f}$ rad/s')

# Frecuencia del notch
ax1.axvline(w0, color='purple', linestyle=':',
            label=r'Notch: $\omega_0=1$ rad/s')

# Puntos de -3 dB
ax1.plot(w1, nivel_corte, 'ro')
ax1.plot(w2, nivel_corte, 'ro')

# Indicar rechazo total
ax1.annotate(r'$|H(j\omega_0)|=0$' '\n' r'$(-\infty$ dB)',
             xy=(w0, -55),
             xytext=(8, 5),
             textcoords='offset points',
             color='purple',
             fontsize=10,
             fontweight='bold')

ax1.set_title('1. Módulo - Filtro Notch')
ax1.set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax1.set_ylabel('Magnitud [dB]')
ax1.set_ylim(-60, 5)
ax1.grid(True, which='both', linestyle='--', alpha=0.7)
ax1.legend(loc='lower left')


# -----------------------------------------------------------------------------
# PANEL 2: Fase
# -----------------------------------------------------------------------------
ax2 = plt.subplot(1, 3, 2)

ax2.semilogx(w, phase, 'b-', linewidth=2,
             label='Respuesta de fase')

ax2.axvline(w0, color='purple', linestyle=':',
            label=r'$\omega_0=1$ rad/s')

ax2.axhline(0, color='gray', linestyle='--',
            label=r'Fase $=0^\circ$')

ax2.set_title('2. Fase - Filtro Notch')
ax2.set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax2.set_ylabel(r'Fase [$^\circ$]')
ax2.grid(True, which='both', linestyle='--', alpha=0.7)
ax2.legend(loc='best')


# -----------------------------------------------------------------------------
# PANEL 3: Polos y Ceros
# -----------------------------------------------------------------------------
ax3 = plt.subplot(1, 3, 3)

polos = sys_notch.poles
ceros = sys_notch.zeros

# Circunferencia de radio w0
theta = np.linspace(0, 2*np.pi, 200)

ax3.plot(w0*np.cos(theta),
         w0*np.sin(theta),
         'k:',
         label=r'$|s|=\omega_0=1$')

# Ejes cartesianos
ax3.axhline(0, color='black', linewidth=1)
ax3.axvline(0, color='black', linewidth=1)

# Polos
ax3.plot(np.real(polos),
         np.imag(polos),
         'rx',
         markersize=10,
         markeredgewidth=2,
         label='Polos')

# Ceros
ax3.plot(np.real(ceros),
         np.imag(ceros),
         'bo',
         markersize=9,
         markerfacecolor='none',
         markeredgewidth=2,
         label='Ceros')

ax3.set_title('3. Diagrama de Polos y Ceros')
ax3.set_xlabel(r'Parte real de $s$')
ax3.set_ylabel(r'Parte imaginaria de $s$')
ax3.grid(True, linestyle='--', alpha=0.7)
ax3.axis('equal')
ax3.legend(loc='upper right')

plt.tight_layout()
plt.show()
```
